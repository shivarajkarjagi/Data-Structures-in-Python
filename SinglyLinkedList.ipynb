### Node Class
A `Node` object represents a single element in the linked list. Each node stores some data and a reference (or link) to the next node in the sequence.

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None  # Pointer to the next node
```

### SinglyLinkedList Class
A `SinglyLinkedList` object manages the entire list. It holds a reference to the `head` node (the first node in the list) and provides methods to perform operations like adding elements, deleting elements, and displaying the list.

```python
class SinglyLinkedList:
    def __init__(self):
        """Initializes an empty SinglyLinkedList with a None head."""
        self.head = None  # Initialize an empty list with no head node

    def append(self, data):
        """Adds a new node with the given data to the end of the list.

        Args:
            data: The data to be stored in the new node.
        """
        new_node = Node(data)
        if not self.head:
            self.head = new_node
            return
        last = self.head
        while last.next:
            last = last.next
        last.next = new_node

    def prepend(self, data):
        """Adds a new node with the given data to the beginning of the list.

        Args:
            data: The data to be stored in the new node.
        """
        new_node = Node(data)
        new_node.next = self.head
        self.head = new_node

    def delete_node(self, key):
        """Deletes the first node found with the given key (data).

        If the key is in the head node, the head is updated. Otherwise, it
        iterates through the list to find and remove the node.

        Args:
            key: The data value of the node to be deleted.
        """
        current = self.head

        # If head node itself holds the key to be deleted
        if current and current.data == key:
            self.head = current.next
            current = None
            return

        prev = None
        while current and current.data != key:
            prev = current
            current = current.next

        # If key was not present in linked list
        if not current:
            print(f"Node with data {key} not found.")
            return

        # Unlink the node from the linked list
        prev.next = current.next
        current = None

    def display(self):
        """Prints the data of all nodes in the list in a 'data -> data' format.
        Prints 'Empty list' if the list contains no nodes.
        """
        elements = []
        current = self.head
        while current:
            elements.append(str(current.data))
            current = current.next
        print(" -> ".join(elements) if elements else "Empty list")

    def get_length(self):
        """Returns the number of nodes in the list.

        Returns:
            int: The total count of nodes in the linked list.
        """
        count = 0
        current = self.head
        while current:
            count += 1
            current = current.next
        return count
```

### Example Usage
Let's see how these classes work together:

```python
# Create a new singly linked list
my_list = SinglyLinkedList()

# Append elements
my_list.append(10)
my_list.append(20)
my_list.append(30)
print("List after appending 10, 20, 30:")
my_list.display()

# Prepend an element
my_list.prepend(5)
print("List after prepending 5:")
my_list.display()

# Delete an element
my_list.delete_node(20)
print("List after deleting 20:")
my_list.display()

# Try to delete a non-existent element
my_list.delete_node(100)
print("List after trying to delete 100:")
my_list.display()

# Delete the head
my_list.delete_node(5)
print("List after deleting 5 (head):")
my_list.display()
```
